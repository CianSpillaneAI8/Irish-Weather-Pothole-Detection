```python
from PIL import Image
import os

# Define input and output directories
input_folder = "/content/drive/MyDrive/Synthetic/Rain/Rain"
output_folder = "/content/drive/MyDrive/Synthetic/Rain/Rain1"

# Create output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Convert PNG to JPG
for filename in os.listdir(input_folder):
    if filename.endswith(".png"):
        png_path = os.path.join(input_folder, filename)
        jpg_path = os.path.join(output_folder, filename.replace(".png", ".jpg"))

        with Image.open(png_path) as img:
            rgb_img = img.convert("RGB")
            rgb_img.save(jpg_path, "JPEG")

        print(f"Converted: {filename} to JPG")

print("Conversion complete!")
```
